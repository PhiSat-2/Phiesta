```python
import numpy as np
import rasterio
import matplotlib.pyplot as plt

def norm(x, percentiles=(2, 98)):
    x = x.astype("float32")
    vals = x[np.isfinite(x)]
    lo, hi = np.percentile(vals, percentiles)
    return np.clip((x - lo) / (hi - lo + 1e-6), 0, 1)

real_rgb = np.dstack([
    norm(ev.get_band("RED")),
    norm(ev.get_band("GREEN")),
    norm(ev.get_band("BLUE")),
])

real_pan = norm(ev.get_band("PAN"))

with rasterio.open(rect["rectified_path"]) as src:
    sim_blue = src.read(1).astype("float32")
    sim_green = src.read(2).astype("float32")
    sim_red = src.read(3).astype("float32")
    sim_pan = src.read(4).astype("float32")

sim_rgb = np.dstack([
    norm(sim_red),
    norm(sim_green),
    norm(sim_blue),
])

sim_pan = norm(sim_pan)

plt.figure(figsize=(8, 8))
plt.imshow(real_rgb)
plt.title("Real ΦSat-2 RGB")
plt.axis("off")
plt.show()

plt.figure(figsize=(8, 8))
plt.imshow(sim_rgb)
plt.title("Simulated ΦSat-2 RGB — catalog rectified + flipped")
plt.axis("off")
plt.show()

plt.figure(figsize=(8, 8))
plt.imshow(real_pan, cmap="gray")
plt.title("Real ΦSat-2 PAN")
plt.axis("off")
plt.show()

plt.figure(figsize=(8, 8))
plt.imshow(sim_pan, cmap="gray")
plt.title("Simulated ΦSat-2 PAN — catalog rectified + flipped")
plt.axis("off")
plt.show()

overlay = np.dstack([
    real_pan,
    sim_pan,
    np.zeros_like(real_pan),
])

plt.figure(figsize=(8, 8))
plt.imshow(overlay)
plt.title("Overlay before refinement: real=red, simulated=green")
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import rasterio
import matplotlib.pyplot as plt

def normalize_u8(x, percentiles=(2, 98)):
    x = x.astype("float32")
    vals = x[np.isfinite(x)]
    lo, hi = np.percentile(vals, percentiles)
    y = np.clip((x - lo) / (hi - lo + 1e-6), 0, 1)
    return (255 * y).astype("uint8")

def resize_max(img, max_size=1800):
    h, w = img.shape[:2]
    scale = min(max_size / max(h, w), 1.0)
    if scale < 1:
        img = cv2.resize(
            img,
            (int(round(w * scale)), int(round(h * scale))),
            interpolation=cv2.INTER_AREA,
        )
    return img, scale

# Real PAN
real_pan = ev.get_band("PAN")

# Rectified simulated PAN = band 4
with rasterio.open(rect["rectified_path"]) as src:
    sim_pan_rect = src.read(4).astype("float32")

real_u8 = normalize_u8(real_pan)
sim_u8 = normalize_u8(sim_pan_rect)

real_small, real_scale = resize_max(real_u8, max_size=1800)
sim_small, sim_scale = resize_max(sim_u8, max_size=1800)

sift = cv2.SIFT_create(nfeatures=8000)

kp_real, des_real = sift.detectAndCompute(real_small, None)
kp_sim, des_sim = sift.detectAndCompute(sim_small, None)

print("keypoints real:", len(kp_real) if kp_real is not None else 0)
print("keypoints sim:", len(kp_sim) if kp_sim is not None else 0)

if des_real is None or des_sim is None:
    raise RuntimeError("No SIFT descriptors found.")

matcher = cv2.BFMatcher(cv2.NORM_L2)
knn = matcher.knnMatch(des_sim, des_real, k=2)

good = []
for m, n in knn:
    if m.distance < 0.75 * n.distance:
        good.append(m)

print("good matches:", len(good))

if len(good) < 4:
    raise RuntimeError(f"Not enough matches: {len(good)}")

src_pts = np.float32([kp_sim[m.queryIdx].pt for m in good]).reshape(-1, 1, 2)
dst_pts = np.float32([kp_real[m.trainIdx].pt for m in good]).reshape(-1, 1, 2)

H_small, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)

if H_small is None or mask is None:
    raise RuntimeError("Homography failed.")

inliers = int(mask.sum())
ratio = inliers / max(len(good), 1)

print("inliers:", inliers)
print("inlier ratio:", ratio)
print("H_small:\n", H_small)

if inliers < 20 or ratio < 0.25:
    raise RuntimeError(
        f"Unreliable refinement: inliers={inliers}, ratio={ratio:.3f}"
    )

warped_sim = cv2.warpPerspective(
    sim_small,
    H_small,
    (real_small.shape[1], real_small.shape[0]),
    flags=cv2.INTER_LINEAR,
)

overlay_before = np.dstack([
    real_small,
    sim_small,
    np.zeros_like(real_small),
])

overlay_after = np.dstack([
    real_small,
    warped_sim,
    np.zeros_like(real_small),
])

plt.figure(figsize=(8, 8))
plt.imshow(overlay_before)
plt.title("Before refinement: real=red, simulated=green")
plt.axis("off")
plt.show()

plt.figure(figsize=(8, 8))
plt.imshow(overlay_after)
plt.title("After SIFT refinement: real=red, simulated=green")
plt.axis("off")
plt.show()
```

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

def norm(x, percentiles=(2, 98)):
    x = x.astype("float32")
    vals = x[np.isfinite(x)]
    lo, hi = np.percentile(vals, percentiles)
    return np.clip((x - lo) / (hi - lo + 1e-6), 0, 1)

aligned_path = "data/triplets/5359/aligned/simulated_rectified_lightglue_aligned.tif"

real_pan = norm(ev.get_band("PAN"))

with rasterio.open(aligned_path) as src:
    aligned_pan = norm(src.read(4).astype("float32"))

valid = aligned_pan > 0

overlay = np.zeros((*real_pan.shape, 3), dtype="float32")
overlay[..., 0] = real_pan
overlay[..., 1] = np.where(valid, aligned_pan, 0)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(real_pan, cmap="gray")
axes[0].set_title("Real ΦSat-2 PAN")
axes[0].axis("off")

axes[1].imshow(aligned_pan, cmap="gray")
axes[1].set_title("Aligned simulated PAN")
axes[1].axis("off")

axes[2].imshow(overlay)
axes[2].set_title("Final overlay: real=red, simulated=green")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
