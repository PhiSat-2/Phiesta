# PyRawPh-Light Quickstart

This notebook shows the main PyRawPh-Light workflows in a minimal way.

## 1. Connect to Insula and load one ΦSat-2 L1 product

```python
from pyrawph import connect_insula

client = connect_insula()
event = client.load_l1("5359")
```

## 2. Inspect the ΦSat-2 product

```python
event.show_event_info()
```

## 3. Display a quick RGB composite

```python
event.show_rgb()
```

## 4. Access individual bands

```python
red = event.get_band("RED")
nir = event.get_band("NIR")
pan = event.get_band("PAN")

print("RED:", red.shape, red.dtype)
print("NIR:", nir.shape, nir.dtype)
print("PAN:", pan.shape, pan.dtype)
```

## 5. Compute a simple spectral index

```python
ndvi = event.index("NDVI")
print("NDVI:", ndvi.shape, ndvi.min(), ndvi.max())
```

## 6. Build a full Sentinel-2 / ΦSat-2 triplet

```python
triplet = event.build_full_sentinel_triplet()
triplet["paths"]
```

## 7. Inspect the generated triplet

```python
event.inspect_full_sentinel_triplet(triplet)
```

## 8. Visualize the triplet

```python
event.show_full_sentinel_triplet(
    triplet,
    save_dir="data/triplets/5359/quicklooks",
)
```

## 9. Use explicit parameters when needed

```python
triplet = event.build_full_sentinel_triplet(
    buffer_km=20.0,
    proxy_target_size=(1024, 1024),
    final_margin_pct=0.15,
    window_days=15,
    max_cloud_cover=20.0,
    verbose=True,
)
```

## 10. Batch processing

```python
from pyrawph.triplets.batch import build_full_sentinel_triplets_batch

batch = build_full_sentinel_triplets_batch(
    client=client,
    product_ids=["5359"],
    output_root="data/triplets",
    verbose=True,
)

batch["rows"]
```

## 11. Read the batch summary as a table

```python
import pandas as pd

df = pd.read_csv(batch["summary_csv"])
df
```

## Notes

The final triplet rasters are designed for pixel-aligned machine learning workflows. The CRS and transform metadata should be treated as grid metadata unless additional geodetic validation is performed.
